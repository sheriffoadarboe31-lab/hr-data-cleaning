```python
#starting data cleaning in python

import pandas as pd
#
#
df = pd.read_csv('Prac_Sql.csv')

display(df.head())




```

```python
print(df.duplicated().sum())
```

```python
# drop a colummn with no data
display(df["Unnamed: 9"])
print(df["Unnamed: 9"].isna().sum()) #"isna" to catch empty entries or values
display(df[df["Unnamed: 9"].notna()]) # catch where is not null
```

```python
# reasigning the data frame "df" to 9 column since the unnaned column has only one entry
df = df.drop(columns=["Unnamed: 9"])
print(df.columns)
```

```python
#strip or trim Names in the name column
df["Name"]=df["Name"].str.strip().str.replace(r"\s+", " ", regex=True) 
#catch the names as string and strip them
print(df["Name"].str.contains("  ").sum())
print(f"Number of duplicate: {df.duplicated().sum()}")
#check for near duplicate 
print(df.duplicated(subset=["Emp_ID"]).sum())
```

```python

unique_only = df.drop_duplicates(subset=["Emp_ID","Department"], keep="first")
display(unique_only.head())
print(unique_only.dtypes) #checking data type before standardizing 
```

```python
#errors ="coerce" put NaN where there is an error in entery
# '\s' - white space '$' dolar sign ',' comma
unique_only["Salary"] = pd.to_numeric(
    #run the inner replace first and then convert to date
    unique_only["Salary"].str.replace(r"[$,\s]", "", regex=True),
    errors="coerce"
)

unique_only["Start_date_cln"] = pd.to_datetime(
    unique_only["Start_date_cln"], format="%d/%m/%y", errors="coerce"
)
#check data type
print(unique_only.dtypes)
```

```python
# getting the number of null shells in each column
print(unique_only.isna().sum())
display(unique_only[unique_only["Emp_ID"].isna()])
unique_only = unique_only.dropna(subset=["Emp_ID"])
# display(unique_only[unique_only["Department"].isna()])

#display(unique_only[unique_only["Salary"].isna()])
print((unique_only["Salary"] ==0).sum())
# set all Salary columns that are 0 or NaN(null) to python null using pd.NA
unique_only["Salary"]=unique_only["Salary"].replace(0, pd.NA)
print(unique_only["Salary"].isna().sum())
display(unique_only[unique_only["Salary"].isna()])
```

```python

unique_only["Salary"]=pd.to_numeric(unique_only["Salary"])
print(unique_only["Salary"].dtype)
print(unique_only["Salary"].isna().sum())
```

```python
# checking the date beyong the expected
print(unique_only["Start_date_cln"].min(), "   " , unique_only["Start_date_cln"].max())
print(unique_only["Start_date_cln"].dt.date.min(), unique_only["Start_date_cln"].dt.date.max())
```

```python
#set empty Gender and department column to 'unknown'
unique_only["Gender"]=unique_only["Gender"].fillna("Unknown")
print(unique_only["Gender"].isna().sum())

# same for Department
unique_only["Department"]=unique_only["Department"].fillna("Unknown")
print(unique_only["Department"].isna().sum())
print(unique_only["Salary"].isna().sum())

# display gender sets number in the gender column
display(unique_only["Gender"].value_counts(dropna=False))
display(unique_only["Department"].value_counts(dropna=False))

#value_counts return sum of unique values like group by for sql
for col in ["Employee_type", "Work_location"]:
    display(unique_only[col].value_counts(dropna=False))
print(unique_only.columns)
```

```python
# rename the file and Save the file
unique_only.to_csv("hr_clean_in_pytn.csv", index=False)
# index = false so the index will not be written as a column by pandas
# Read the file back into a variable
hr_clean_in_pytn = pd.read_csv("hr_clean_in_pytn.csv")


display(hr_clean_in_pytn.head())
```
